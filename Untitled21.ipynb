### 3.5 Standardize units and numeric types
- **Capacity:** converted to numeric, then `capacity_gw` = MW / 1000 (GW values kept as they are).
- **Generation:** converted to numeric, then `generation_twh` = GWh / 1000 (TWh values kept).
- **ISO code:** trimmed and upper-cased so it can be used as a merge key.
- **Consumption:** `consumption_twh` = GWh / 1000, or PJ x 0.277778 (1 PJ = 0.277778 TWh).
- **Boundary checks:** negative capacity, generation or consumption values are set to `NaN`.
- `describe()` is used as a sanity check on the converted columns.

**Notes:**
- Values stored as text such as `1,234.5` or `45.2 MW` cannot be parsed and become `NaN`.
- Rows with a missing unit are left unconverted.
- The extreme maxima in `describe()` (e.g. capacity of about 138,453 GW) show that unit gaps and outliers still remain at this point.
- Missing ISO codes became the text `NAN`, so those rows cannot match the Ember data later.

```python
# 1. Clean Capacity and convert MW to GW
df_raw["installed_capacity"] = pd.to_numeric(df_raw["installed_capacity"].replace("-", np.nan), errors="coerce")
df_raw["capacity_gw"] = np.where(df_raw["capacity_unit"] == "MW", df_raw["installed_capacity"] / 1000.0, df_raw["installed_capacity"])
```

```python
# 2. Clean Generation and convert GWh to TWh
df_raw["generation"] = pd.to_numeric(df_raw["generation"].replace("-", np.nan), errors="coerce")
df_raw["generation_twh"] = np.where(df_raw["generation_unit"] == "GWh", df_raw["generation"] / 1000.0, df_raw["generation"])
```

```python
# 3. Clean ISO code for reliable dataset merging
df_raw["iso_code"] = df_raw["iso_code"].astype(str).str.strip().str.upper().replace("-", np.nan)
```

```python
# 3. Clean Electricity Consumption and convert all units to TWh
df_raw["electricity_consumption"] = pd.to_numeric(df_raw["electricity_consumption"].replace("-", np.nan), errors="coerce")
df_raw["consumption_twh"] = df_raw["electricity_consumption"]
df_raw["consumption_twh"] = np.where(df_raw["consumption_unit"] == "GWh", df_raw["electricity_consumption"] / 1000.0, df_raw["consumption_twh"])
df_raw["consumption_twh"] = np.where(df_raw["consumption_unit"] == "PJ", df_raw["electricity_consumption"] * 0.277778, df_raw["consumption_twh"])
```

```python
# Boundary checks: negative generation or capacity set to NaN
df_raw.loc[df_raw["capacity_gw"] < 0, "capacity_gw"] = np.nan
df_raw.loc[df_raw["generation_twh"] < 0, "generation_twh"] = np.nan
df_raw.loc[df_raw["consumption_twh"] < 0, "consumption_twh"] = np.nan
```

### 4.7 Build Ember benchmark table
- `ember_demand_twh`: the Ember `Demand` rows (national electricity consumption).
- `ember_emissions_mtco2` and `ember_emissions_intensity`: the `Total generation` rows (power-sector emissions and gCO2e/kWh).
- The two tables are outer-merged on `iso_code` and `year` into `df_ember_bench`.

```python
# 1. Clean country-level electricity demand benchmark (consumption in TWh)
ember_demand = df_ember_raw[df_ember_raw["Electricity source"] == "Demand"][["iso_code", "year", "generation_twh"]].copy()
ember_demand.columns = ["iso_code", "year", "ember_demand_twh"]

# 2. Clean country-level power sector carbon emissions and emission intensity
ember_emissions = df_ember_raw[df_ember_raw["Electricity source"] == "Total generation"][["iso_code", "year", "emissions_mtco2", "emissions_intensity"]].copy()
ember_emissions.columns = ["iso_code", "year", "ember_emissions_mtco2", "ember_emissions_intensity"]

# 3. Merge Ember benchmarks together
df_ember_bench = pd.merge(ember_demand, ember_emissions, on=["iso_code", "year"], how="outer")
df_ember_bench.head()
```

## 5. Integration and Feature Engineering
- `df_clean` is left-merged with `df_ember_bench` on `iso_code` and `year`, giving `df_master`.
- `population_millions` and `gdp_billion_usd` are converted to numeric.
- **New feature `gdp_per_capita_usd`** = GDP (USD) / population.
- **New feature `clean_share_calculated_%`**: each row's generation as a percentage of the total generation of its country-year. Despite the name, this is the share of that *source* in the country's total, not a clean-energy share.
- Final shape: 10,401 rows x 31 columns.

**Note:** rows with a missing ISO code (stored as `NAN`) cannot match the Ember data and get empty benchmark values.

```python
# Merge primary clean energy data with secondary Ember benchmarks on iso_code and year
df_master = pd.merge(df_clean, df_ember_bench, on=["iso_code", "year"], how="left")

# Convert population and GDP to numeric
df_master["population_millions"] = pd.to_numeric(df_master["population_millions"].replace("-", np.nan), errors="coerce")
df_master["gdp_billion_usd"] = pd.to_numeric(df_master["gdp_billion_usd"].replace("-", np.nan), errors="coerce")

# Create derived feature: GDP per Capita (USD)
df_master["gdp_per_capita_usd"] = (df_master["gdp_billion_usd"] * 1e9) / (df_master["population_millions"] * 1e6)

# Create derived feature: Clean Generation Ratio per Country-Year
country_year_totals = df_master.groupby(["iso_code", "year"])["generation_twh"].transform("sum")
df_master["clean_share_calculated_%"] = (df_master["generation_twh"] / country_year_totals) * 100

print("Master Integrated Shape:", df_master.shape)
df_master.head()
```

## 6. Analysis and Visualisation
### 6.1 Yearly renewable vs non-renewable generation
- Generation (TWh) is summed by year and source type.
- `Renewable_Share_%` is renewable generation divided by total generation.
- `Renewable_YoY_Growth_%` is the year-over-year percentage change in renewable generation.

**Note:** the yearly values swing sharply (e.g. a 1.7% share in 2017 vs 75.6% in 2016) because the unconverted and outlier values described in step 3.5 dominate the sums. Treat these figures with caution until those values are handled.

```python
# Aggregate generation by year and source type
yearly_gen = df_master.groupby(["year", "clean_source_type"])["generation_twh"].sum().unstack().fillna(0)

# Calculate annual renewable share %
yearly_gen["Renewable_Share_%"] = (yearly_gen["Renewable"] / (yearly_gen["Renewable"] + yearly_gen["Non Renewable"])) * 100

# Calculate Year-over-Year (YoY) Growth %
yearly_gen["Renewable_YoY_Growth_%"] = yearly_gen["Renewable"].pct_change() * 100

yearly_gen.tail(10).round(2)
```

### Chart 1: Stacked bar chart, global generation mix by source (2000-2022)
Annual generation (TWh) stacked by energy source to show how the mix changes over time.

```python
# Group generation by year and energy source
mix_table = df_master.groupby(["year", "clean_source"])["generation_twh"].sum().unstack().fillna(0)

mix_table.plot(kind="bar", stacked=True, figsize=(11, 5), colormap="tab10")
plt.title("Chart 1: Global Electricity Generation Mix by Source (2000 - 2022)")
plt.xlabel("Year")
plt.ylabel("Electricity Generation (TWh)")
plt.legend(title="Energy Source", bbox_to_anchor=(1.02, 1), loc="upper left")
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

### Chart 2: Time series (dual axis), renewable generation vs grid carbon intensity
Yearly renewable generation (green, left axis) plotted against the average Ember carbon intensity in gCO2e/kWh (red, right axis).

```python
# Yearly total renewable generation vs average Ember carbon intensity
yearly_emissions = df_master.groupby("year").agg({
    "generation_twh": lambda x: x[df_master.loc[x.index, "clean_source_type"] == "Renewable"].sum(),
    "ember_emissions_intensity": "mean"
}).reset_index()

fig, ax1 = plt.subplots(figsize=(10, 5))
color = "tab:green"
ax1.set_xlabel("Year")
ax1.set_ylabel("Renewable Generation (TWh)", color=color)
ax1.plot(yearly_emissions["year"], yearly_emissions["generation_twh"], color=color, marker="o", linewidth=2)
ax1.tick_params(axis="y", labelcolor=color)

ax2 = ax1.twinx()
color = "tab:red"
ax2.set_ylabel("Avg Carbon Intensity (gCO2e/kWh)", color=color)
ax2.plot(yearly_emissions["year"], yearly_emissions["ember_emissions_intensity"], color=color, marker="s", linestyle="--", linewidth=2)
ax2.tick_params(axis="y", labelcolor=color)

plt.title("Chart 2: Growth in Renewable Generation vs. Grid Carbon Intensity")
plt.grid(True, linestyle=":", alpha=0.6)
plt.tight_layout()
plt.show()
```

Renewable share of generation for each region in 2022, sorted from lowest to highest.

**Note:** region names were not standardized in the cleaning steps, so label variants (e.g. `S. Asia`, `East Asia and Pacific`) appear as separate bars, and rows with a missing region are excluded.

```python
# Filter for year 2022 and valid regions
df_2022 = df_master[(df_master["year"] == 2022) & (df_master["region"].notna())]
reg_mix = df_2022.groupby(["region", "clean_source_type"])["generation_twh"].sum().unstack().fillna(0)
reg_mix["Renewable_Share_%"] = (reg_mix["Renewable"] / (reg_mix["Renewable"] + reg_mix["Non Renewable"])) * 100
reg_mix = reg_mix.sort_values(by="Renewable_Share_%", ascending=True)

plt.figure(figsize=(10, 5))
plt.barh(reg_mix.index, reg_mix["Renewable_Share_%"], color="teal", edgecolor="black")
plt.title("Chart 3: Renewable Energy Share of Generation by Region (2022)")
plt.xlabel("Renewable Share (%)")
plt.ylabel("Region")
plt.xlim(0, 100)
plt.grid(axis="x", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

### Chart 4: Scatter plot, GDP per capita vs installed renewable capacity (2021)
One point per country: mean GDP per capita against total installed renewable capacity (GW).

```python
# Aggregate country-level metrics for 2021
df_scatter = df_master[df_master["year"] == 2021].groupby("country").agg({
    "gdp_per_capita_usd": "mean",
    "capacity_gw": lambda x: x[df_master.loc[x.index, "clean_source_type"] == "Renewable"].sum()
}).dropna()

plt.figure(figsize=(9, 5))
plt.scatter(df_scatter["gdp_per_capita_usd"], df_scatter["capacity_gw"], color="darkblue", alpha=0.7)
plt.title("Chart 4: GDP per Capita vs. Installed Renewable Capacity (2021)")
plt.xlabel("GDP per Capita (USD)")
plt.ylabel("Installed Renewable Capacity (GW)")
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```
